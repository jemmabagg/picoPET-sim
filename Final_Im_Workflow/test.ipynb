```python
import numpy as np
import matplotlib.pyplot as plt 
from skimage.transform import iradon, radon
```

```python
N = 128
x = np.linspace(-1, 1, N)
y = np.linspace(-1, 1, N)

X, Y = np.meshgrid(x, y)

R = 0.8
circle = X**2 + Y**2 <= R**2

image = np.zeros((N, N))
image[circle] = 1
```

```python
plt.imshow(image)
plt.title("Uniform Circle")
plt.colorbar()
plt.show()
```

```python
theta = np.linspace(0, 180, 180, endpoint=False)

sinogram = radon(image, theta=theta, circle=True)
```

```python
plt.imshow(sinogram, aspect="auto")
plt.xlabel("theta")
plt.ylabel("s")
plt.colorbar()
plt.show()
```
